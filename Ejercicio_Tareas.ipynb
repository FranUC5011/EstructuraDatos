```python
# Elaborado por: Francisco Urrutia Calero - 25402171
# GRUPO SI2DB

# *********************** C O N D I C I O N A L E S ****************************

# Ejercicio 1: Impuestos:
print("\n--- EJERCICIO 1: VERIFICAR IMPUESTOS ---") # Esto es importante

edad = int(input("Ingresa tu edad: "))
ingresos = float(input("Ingresa tus ingresos mensuales en cordobas: "))

if edad >= 16 and ingresos >= 1000: # Esto creo que significa que tienes que pagar impuestos
    print("Debes tributar el impuesto")  # Espero que esto sea correcto
else:  # A veces confundo else con otra cosa
    print("No debes tributar el impuesto") # No sé si esto está bien

# Ejercicio 2: Tramos impositivos:
print("\n--- EJERCICIO 2: TRAMOS IMPOSITIVOS ---")

renta = float(input("Ingresa tu renta anual en euros: ")) # Cuidado aquí

# Aquí estaba confundido con los tipos
if renta < 10000:
    tipo = "5%"
elif renta < 20000:
    tipo = "15%"
elif renta < 35000:
    tipo = "20%"
elif renta < 60000:
    tipo = "30%"
else:
    tipo = "45%"  # Ojalá esto funcione

# Aquí imprimo el tipo
print("Tu tipo impositivo es: " + tipo)

# Ejercicio 3: Descuento por edad:
print("\n--- EJERCICIO 3: DESCUENTO POR EDAD ---")
edad = int(input("Ingresa tu edad: ")) # Por favor ingresa bien
precio = float(input("Ingresa el precio del producto: ")) # Esto debe ser un número

if edad < 18: # menores de edad
    descuento = 0.10  # organizar mis ideas
elif edad >= 65:
    descuento = 0.15
else:
    descuento = 0.0  # Espero que no se me olvide algo

precio_final = precio * (1 - descuento)
print("Precio final a pagar: $" + str(precio_final))  # ¡Esto me costó entender!

# Ejercicio 4: Clasificacion de notas:
print("\n--- EJERCICIO 4: CLASIFICACION DE NOTAS ---")

nota = float(input("Ingresa tu nota (0-100): "))

# Las notas a veces son difíciles de clasificar
if nota >= 90 and nota <= 100:
    calificacion = "Excelente"
elif nota >= 70 and nota <= 89:
    calificacion = "Bueno"
elif nota >= 60 and nota <= 69:
    calificacion = "Regular"
elif nota >= 0 and nota < 60:
    calificacion = "Reprobado"
else:
    calificacion = "Nota invalida"  # Esto es importante

print("Tu calificacion es: " + calificacion) # Espero que no sea muy tarde

# Ejercicio 5: Mayor de tres numeros:
print("\n--- EJERCICIO 5: MAYOR DE TRES NUMEROS ---")

num1 = float(input("Ingresa el primer numero: "))  # Debo tener cuidado aquí
num2 = float(input("Ingresa el segundo numero: "))
num3 = float(input("Ingresa el tercer numero: "))

# Determinando el mayor
if num1 >= num2 and num1 >= num3:
    mayor = num1
elif num2 >= num1 and num2 >= num3:
    mayor = num2
else:
    mayor = num3

print("El numero mayor es: " + str(mayor))  # ¡Esto me tomó tiempo!

# Ejercicio 6: Clasificacion de triangulos:
print("\n--- EJERCICIO 6: CLASIFICACION DE TRIANGULOS ---")

lado1 = float(input("Ingresa el primer lado: "))
lado2 = float(input("Ingresa el segundo lado: "))
lado3 = float(input("Ingresa el tercer lado: "))

# Clasificando triángulos
if lado1 == lado2 and lado2 == lado3:
    tipo = "Equilatero"
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    tipo = "Isosceles"
else:
    tipo = "Escaleno"

print("El triangulo es: " + tipo)  # Espero que esto sea útil

# Ejercicio 7: Simulador de cajero:
print("\n--- EJERCICIO 7: SIMULADOR DE CAJERO ---")

saldo = 5000

monto = float(input("Cuantos deseas retirar? ")) # cuidado con esto

if monto > saldo:
    print("Fondos insuficientes")
else:
    saldo = saldo - monto  # Restando del saldo
    print("Retiro exitoso. Nuevo saldo: $" + str(saldo))  # ¡Espero que funcione!

# *********************** B U C L E S ****************************

# Ejercicio 8:
print("\n--- EJERCICIO 8: CUENTA ATRAS ---")

numero = int(input("Ingresa un numero entero positivo: "))

if numero < 0:  # No se pueden negativos
    print("El numero debe ser positivo")
else:
    print("Cuenta atras: ")
    for i in range(numero, -1, -1):  # Esto es fácil
        if i == 0:
            print(i)
        else:
            print(i, end=", ")  # Me gusta hacer esto

# Ejercicio 9:
print("\n--- EJERCICIO 9: TABLA DE MULTIPLICAR ---")

print("TABLAS DE MULTIPLICAR DEL 1 AL 10")
print("-"*50)

for i in range(1, 11):
    print("")
    print("Tabla del " + str(i) + ":")
    for j in range(1, 11):
        resultado = i * j
        print(str(i) + " x " + str(j) + " = " + str(resultado))  # Esto suena correcto

# Ejercicio 10:
print("\n--- EJERCICIO 10: ADIVINA LA CONTRASEÑA ---")

contrasena_correcta = "1234"
intentos = 0
max_intentos = 5
acerto = False

# Aquí ten cuidado con los intentos
while intentos < max_intentos:
    contrasena = input("Ingresa tu contrasena: ")
    intentos += 1  # Así es como sumo

    if contrasena == contrasena_correcta:  # Éxito
        print("Contrasena correcta")
        acerto = True
        break
    else:
        print("Contrasena incorrecta")  # Esto pasó antes!

if acerto == False:
    print("Demasiados intentos fallidos")  # Lamentable

# Ejercicio 11: Contar letras en una frase:
print("\n--- EJERCICIO 11: CONTAR LETRAS EN UNA FRASE ---")

frase = input("Ingresa una frase: ")
letra = input("Ingresa una letra a buscar: ")

contador = 0  # Contador para letras

for caracter in frase:  # Aquí contamos las letras
    if caracter == letra:
        contador += 1  # Sumando uno si coincide

print("La letra '" + letra + "' aparece " + str(contador) + " veces en la frase")

# Ejercicio 12: Mostrar letras al reves:
print("\n--- EJERCICIO 12: MOSTRAR LETRAS AL REVÉS ---")

palabra = input("Ingresa una palabra: ")

print("La palabra al revés es: ", end="")

# Aquí usando un ciclo para invertir la palabra
for i in range(len(palabra) - 1, -1, -1):
    print(palabra[i], end="")

print()  # Fin de la línea
```
